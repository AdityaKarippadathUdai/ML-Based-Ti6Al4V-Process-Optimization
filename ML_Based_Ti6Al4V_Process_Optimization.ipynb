# Check missing values

```python
missing_values = df.isnull().sum()

missing_values
```

```python
if missing_values.sum() == 0:
    print("No missing values found.")
else:
    print("Missing values detected.")
```

# Missing-value percentage

```python
missing_percentage = (
    df.isnull().sum() / len(df) * 100
).sort_values(ascending=False)

missing_percentage
```

# Check duplicate rows

```python
duplicate_count = df.duplicated().sum()

print(f"Number of duplicate rows: {duplicate_count}")
```

# Number of unique values

```python
unique_values = pd.DataFrame({
    "Column": df.columns,
    "Unique Values": [df[col].nunique() for col in df.columns]
})

unique_values
```

# Check Alloy column

```python
print("Number of unique Alloy values:", df["Alloy"].nunique())
print("Number of rows:", len(df))

df["Alloy"].value_counts().sort_index().head(20)
```

```python
if df["Alloy"].nunique() == len(df):
    print(
        "Alloy contains a unique value for every row. "
        "It should be investigated as an identifier rather than "
        "automatically used as a predictive feature."
    )
```

# Reference analysis

```python
print("Number of unique references:", df["Reference"].nunique())

reference_counts = df["Reference"].value_counts()

reference_counts
```

# Samples per reference

```python
plt.figure(figsize=(10, 5))

plt.hist(
    reference_counts.values,
    bins=15
)

plt.xlabel("Number of Samples")
plt.ylabel("Number of References")
plt.title("Distribution of Samples per Reference")

plt.tight_layout()
plt.show()
```

# Define process parameters

```python
process_features = [
    "Powder Size (μm)",
    "Laser Spot (μm)",
    "Laser Power (W)",
    "Scanning Speed (mm/s)",
    "Hatch Distance (μm)",
    "Layer Thickness (μm)"
]

geometry_features = [
    "Gauge Area (mm²)",
    "Gauge Length (mm)"
]

targets = [
    "UTS (MPa)",
    "YS (MPa)",
    "EF (%)"
]

print("Process parameters:")
print(process_features)

print("\nGeometry variables:")
print(geometry_features)

print("\nTargets:")
print(targets)
```

# Process parameter summary

```python
df[process_features].describe().T
```

# Target summary

```python
df[targets].describe().T
```

# Distribution of process parameters

```python
for column in process_features:

    plt.figure(figsize=(8, 5))

    plt.hist(
        df[column],
        bins=20
    )

    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.title(f"Distribution of {column}")

    plt.tight_layout()
    plt.show()
```

# Distribution of targets

```python
for column in targets:

    plt.figure(figsize=(8, 5))

    plt.hist(
        df[column],
        bins=20
    )

    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.title(f"Distribution of {column}")

    plt.tight_layout()
    plt.show()
```

# Boxplots of process parameters

```python
for column in process_features:

    plt.figure(figsize=(8, 4))

    plt.boxplot(
        df[column],
        vert=False
    )

    plt.xlabel(column)
    plt.title(f"Boxplot of {column}")

    plt.tight_layout()
    plt.show()
```

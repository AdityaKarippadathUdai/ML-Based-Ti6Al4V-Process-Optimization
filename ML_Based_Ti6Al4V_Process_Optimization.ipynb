# Import libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import (
    RBF,
    ConstantKernel,
    WhiteKernel
)

from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error
)

print("Libraries imported successfully")
```

# Load dataset

```python
FILE_PATH = "/content/1-s2.0-S2214860424003877-mmc1.xlsx"

df = pd.read_excel(FILE_PATH)

print("Dataset shape:", df.shape)

df.head()
```

# Check Alloy

```python
print("Alloy unique values:", df["Alloy"].nunique())
print("Number of rows:", len(df))

print("\nFirst 20 Alloy values:")
print(df["Alloy"].head(20).tolist())
```

# Verify the ML inputs

```python
features = [
    "Powder Size (μm)",
    "Laser Spot (μm)",
    "Laser Power (W)",
    "Scanning Speed (mm/s)",
    "Hatch Distance (μm)",
    "Layer Thickness (μm)",
    "Gauge Area (mm²)",
    "Gauge Length (mm)"
]

targets = [
    "UTS (MPa)",
    "YS (MPa)",
    "EF (%)"
]

X = df[features].copy()
y = df[targets].copy()

print("Input shape:", X.shape)
print("Target shape:", y.shape)

display(X.head())
display(y.head())
```

# Define Features and Targets

```python
feature_cols = [
    "Powder Size (μm)",
    "Laser Spot (μm)",
    "Laser Power (W)",
    "Scanning Speed (mm/s)",
    "Hatch Distance (μm)",
    "Layer Thickness (μm)",
    "Gauge Area (mm²)",
    "Gauge Length (mm)"
]

target_cols = [
    "UTS (MPa)",
    "YS (MPa)",
    "EF (%)"
]

X = df[feature_cols]
y = df[target_cols]

print("Input shape :", X.shape)
print("Target shape:", y.shape)
```

# Train/Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

# Define Models

```python
models = {

    "GPR": Pipeline([
        ("scaler", StandardScaler()),

        ("model", GaussianProcessRegressor(
            kernel=(
                ConstantKernel(1.0)
                * RBF(1.0)
                + WhiteKernel(1e-5)
            ),
            normalize_y=True,
            n_restarts_optimizer=3,
            random_state=42
        ))
    ]),

    "SVR": Pipeline([
        ("scaler", StandardScaler()),

        ("model", SVR(
            kernel="rbf",
            C=100,
            gamma="scale",
            epsilon=0.1
        ))
    ]),

    "Random Forest": RandomForestRegressor(
        n_estimators=300,
        max_depth=None,
        min_samples_split=2,
        min_samples_leaf=1,
        random_state=42,
        n_jobs=-1
    )
}

print("Models:")
for name in models:
    print("-", name)
```

# Train All Models for UTS

```python
target = "UTS (MPa)"

y_train_target = y_train[target]
y_test_target = y_test[target]

uts_models = {}

for name, model in models.items():

    print(f"Training {name}...")

    model.fit(X_train, y_train_target)

    uts_models[name] = model

print("UTS training completed")
```

# Evaluate UTS

```python
uts_results = []

for name, model in uts_models.items():

    predictions = model.predict(X_test)

    r2 = r2_score(y_test_target, predictions)
    rmse = np.sqrt(mean_squared_error(y_test_target, predictions))
    mae = mean_absolute_error(y_test_target, predictions)

    uts_results.append({
        "Model": name,
        "R²": r2,
        "RMSE": rmse,
        "MAE": mae
    })

uts_results_df = pd.DataFrame(uts_results)

uts_results_df.sort_values(
    by="R²",
    ascending=False
)
```

# Train All Models for YS

```python
target = "YS (MPa)"

y_train_target = y_train[target]
y_test_target = y_test[target]

ys_models = {}

for name, model in models.items():

    print(f"Training {name}...")

    model.fit(X_train, y_train_target)

    ys_models[name] = model

print("YS training completed")
```

# Evaluate YS

```python
ys_results = []

for name, model in ys_models.items():

    predictions = model.predict(X_test)

    r2 = r2_score(y_test_target, predictions)
    rmse = np.sqrt(mean_squared_error(y_test_target, predictions))
    mae = mean_absolute_error(y_test_target, predictions)

    ys_results.append({
        "Model": name,
        "R²": r2,
        "RMSE": rmse,
        "MAE": mae
    })

ys_results_df = pd.DataFrame(ys_results)

ys_results_df.sort_values(
    by="R²",
    ascending=False
)
```

# Train All Models for EF

```python
target = "EF (%)"

y_train_target = y_train[target]
y_test_target = y_test[target]

ef_models = {}

for name, model in models.items():

    print(f"Training {name}...")

    model.fit(X_train, y_train_target)

    ef_models[name] = model

print("EF training completed")
```

# Evaluate EF

```python
ef_results = []

for name, model in ef_models.items():

    predictions = model.predict(X_test)

    r2 = r2_score(y_test_target, predictions)
    rmse = np.sqrt(mean_squared_error(y_test_target, predictions))
    mae = mean_absolute_error(y_test_target, predictions)

    ef_results.append({
        "Model": name,
        "R²": r2,
        "RMSE": rmse,
        "MAE": mae
    })

ef_results_df = pd.DataFrame(ef_results)

ef_results_df.sort_values(
    by="R²",
    ascending=False
)
```

# Combined Results

```python
all_results = pd.concat([
    uts_results_df.assign(Target="UTS"),
    ys_results_df.assign(Target="YS"),
    ef_results_df.assign(Target="EF")
], ignore_index=True)

all_results = all_results[
    ["Target", "Model", "R²", "RMSE", "MAE"]
]

all_results.sort_values(
    by=["Target", "R²"],
    ascending=[True, False]
)
```

# R² Comparison

```python
r2_table = all_results.pivot(
    index="Model",
    columns="Target",
    values="R²"
)

r2_table
```

# Plot R²

```python
r2_table.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.ylabel("R²")
plt.xlabel("Model")
plt.title("Model R² Comparison")
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
target_models = {
    "UTS": (y_test["UTS (MPa)"], uts_models),
    "YS": (y_test["YS (MPa)"], ys_models),
    "EF": (y_test["EF (%)"], ef_models)
}

for target_name, (actual, trained_models) in target_models.items():

    for model_name, model in trained_models.items():

        predictions = model.predict(X_test)

        plt.figure(figsize=(6, 6))

        plt.scatter(actual, predictions)

        min_val = min(actual.min(), predictions.min())
        max_val = max(actual.max(), predictions.max())

        plt.plot(
            [min_val, max_val],
            [min_val, max_val],
            linestyle="--"
        )

        plt.xlabel(f"Actual {target_name}")
        plt.ylabel(f"Predicted {target_name}")

        plt.title(
            f"{model_name} — {target_name}"
        )

        plt.tight_layout()
        plt.show()
```

# Train GPR, SVR and Random Forest inside each GMM cluster

```python
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import (
    RBF,
    ConstantKernel,
    WhiteKernel
)
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error


targets = ["UTS (MPa)", "YS (MPa)", "EF (%)"]

results = []

for target in targets:

    print(f"\n{'='*60}")
    print(f"Target: {target}")
    print(f"{'='*60}")

    y_train_target = y_train[target]
    y_test_target = y_test[target]

    for model_name in ["GPR", "SVR", "Random Forest"]:

        predictions = np.full(len(X_test), np.nan)

        print(f"\nModel: {model_name}")

        # Train one model for every cluster
        for cluster in range(3):

            train_mask = train_clusters == cluster
            test_mask = test_clusters == cluster

            X_cluster_train = X_train.iloc[train_mask]
            y_cluster_train = y_train_target.iloc[train_mask]

            X_cluster_test = X_test.iloc[test_mask]

            print(
                f"Cluster {cluster}: "
                f"train={len(X_cluster_train)}, "
                f"test={len(X_cluster_test)}"
            )

            # Skip if cluster has no training samples
            if len(X_cluster_train) == 0:
                continue

            # -------------------------
            # GPR
            # -------------------------
            if model_name == "GPR":

                kernel = (
                    ConstantKernel(1.0)
                    * RBF(length_scale=1.0)
                    + WhiteKernel(noise_level=1.0)
                )

                model = GaussianProcessRegressor(
                    kernel=kernel,
                    normalize_y=True,
                    n_restarts_optimizer=3,
                    random_state=42
                )

                # GPR benefits from scaled features
                scaler = StandardScaler()

                X_cluster_train_scaled = scaler.fit_transform(
                    X_cluster_train
                )

                X_cluster_test_scaled = scaler.transform(
                    X_cluster_test
                )

                model.fit(
                    X_cluster_train_scaled,
                    y_cluster_train
                )

                if len(X_cluster_test) > 0:
                    predictions[test_mask] = model.predict(
                        X_cluster_test_scaled
                    )

            # -------------------------
            # SVR
            # -------------------------
            elif model_name == "SVR":

                scaler = StandardScaler()

                X_cluster_train_scaled = scaler.fit_transform(
                    X_cluster_train
                )

                X_cluster_test_scaled = scaler.transform(
                    X_cluster_test
                )

                model = SVR(
                    kernel="rbf",
                    C=100,
                    gamma="scale",
                    epsilon=0.1
                )

                model.fit(
                    X_cluster_train_scaled,
                    y_cluster_train
                )

                if len(X_cluster_test) > 0:
                    predictions[test_mask] = model.predict(
                        X_cluster_test_scaled
                    )

            # -------------------------
            # Random Forest
            # -------------------------
            elif model_name == "Random Forest":

                model = RandomForestRegressor(
                    n_estimators=300,
                    max_depth=None,
                    min_samples_split=2,
                    min_samples_leaf=1,
                    random_state=42,
                    n_jobs=-1
                )

                model.fit(
                    X_cluster_train,
                    y_cluster_train
                )

                if len(X_cluster_test) > 0:
                    predictions[test_mask] = model.predict(
                        X_cluster_test
                    )

        # -------------------------
        # Evaluate complete test set
        # -------------------------

        valid = ~np.isnan(predictions)

        y_true = y_test_target.values[valid]
        y_pred = predictions[valid]

        r2 = r2_score(y_true, y_pred)
        rmse = np.sqrt(mean_squared_error(y_true, y_pred))
        mae = mean_absolute_error(y_true, y_pred)

        results.append({
            "Target": target,
            "Model": model_name,
            "R²": r2,
            "RMSE": rmse,
            "MAE": mae
        })

        print(f"R²   : {r2:.4f}")
        print(f"RMSE : {rmse:.4f}")
        print(f"MAE  : {mae:.4f}")
```
